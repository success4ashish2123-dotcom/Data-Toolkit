8.What does the term “vectorized operation” mean in NumPy.

Ans.
A vectorized operation in NumPy means performing a mathematical or computational operation on an entire array or multiple array elements at once, without explicitly writing a Python loop for each element.

In simple words: Instead of applying an operation to each element manually using a loop, we apply the operation directly to the entire NumPy array.

1.  How do you create a 2D NumPy array and calculate the sum of each row.

Ans. In NumPy, we can create a 2D array using the np.array() function and calculate the sum of each row using the np.sum() function with the axis=1 parameter.

A 2D array consists of rows and columns, similar to a table.
- axis=0 calculates the sum of each column.
- axis=1 calculates the sum of each row.

import numpy as np

# Create a 2D NumPy array
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Calculate the sum of each row
row_sum = np.sum(arr, axis=1)

# Display the results
print("2D Array:")
print(arr)

print("Sum of Each Row:")
print(row_sum)

Ans.

A correlation matrix shows how strongly numerical variables are related to each other.

- Study hours
- Exam marks
- Sleep hours

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

    'Study_Hours': [1, 2, 3, 4, 5, 6],

    linewidths=0.5
